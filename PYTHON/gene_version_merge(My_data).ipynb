```python
mapping_df=pd.read_csv("/staging/biology/jane0528/NMOSD/scRNA/Dataset/GSE164378_v94_to_v44.csv")
mapping_dict = dict(zip(mapping_df["HugoSymbol_v94"], mapping_df["HugoSymbol_merge"]))
```

```python
#讀取 GSE164378 features.tsv.gz
features = pd.read_csv("/staging/biology/jane0528/NMOSD/scRNA/Dataset/GSE164378_features.tsv", sep="\t", header=None, names=["HUGO1", "HUGO2", "TYPE"])
print("原始 features 行數:", len(features))

# 替換
features["HUGO1"] = features["HUGO1"].map(mapping_dict).fillna(features["HUGO1"])
features["HUGO2"] = features["HUGO2"].map(mapping_dict).fillna(features["HUGO2"])
```

```python
# 輸出新的 features 檔案
features.to_csv("/staging/biology/jane0528/NMOSD/scRNA/Dataset/GSE164378_features_renamed(v44).tsv", sep="\t", index=False, header=False)
```
