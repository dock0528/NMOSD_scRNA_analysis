### 合併gencode v32 & v44

```python
ref_v32_genes = pd.read_csv("./Dataset/mapping_v32.csv",header=None,names=["HugoSymbol","ENSG"])
ref_v32_genes
```

```python
ref_v32_genes["ENSG"] = ref_v32_genes["ENSG"].str.split(".").str[0]
ref_v32_genes
```

```python
ref_v44_genes = pd.read_csv("./Dataset/features_v44.tsv",header=None,names=["ENSG","HugoSymbol","Type"],sep="\t").loc[:,["ENSG","HugoSymbol"]]
ref_v44_genes
```

```python
#Gencode v44 duplicate check
dup_v44_ENSG = ref_v44_genes[ref_v44_genes["ENSG"].duplicated(keep=False)]
dup_v44_hugo = ref_v44_genes[ref_v44_genes["HugoSymbol"].duplicated(keep=False)]

print("v44 Ensembl ID duplicates counts:",dup_v44_ENSG.shape[0])
print("v44 Ensembl ID duplicates:",dup_v44_ENSG)
print("----------------------------------------------")
print("v44 Hugo Symbol duplicates counts:",dup_v44_hugo.shape[0])
print("v44 Hugo Symbol duplicates :\n",dup_v44_hugo)

```

```python
#Gencode v32 duplicate check
dup_v32 = ref_v32_genes[ref_v32_genes["ENSG"].duplicated(keep=False)]
print("V32 duplicates:\n", dup_v32)
```

```python
#NMOSD features
nmosd_hugo=pd.read_csv("./Dataset/nmosd/mtx_nmosd/Origin_DATA/nmosd_features_ori.tsv",header=None,names=["HugoSymbol"]) 
nmosd_v32_genes=nmosd_hugo.merge(ref_v32_genes,on="HugoSymbol",how="left")
print("NMOSD original Hugo Symbol counts:",nmosd_hugo.shape[0])
print("NMOSD mapping to genecode v32 gene counts:",nmosd_v32_genes.shape[0])
nmosd_v32_genes
```

```python
nmosd_v32_genes.to_csv("./Dataset/nmosd_v32_HUGO_map_ENSG.csv",index=False)
```

```python
# NMOSD_v32 和 Control_v44 features merge
merged_genes_nmosd_control=ref_v44_genes.merge(nmosd_v32_genes,on="ENSG",how="inner",suffixes=('_v44', '_v32'))
merged_genes_nmosd_control
```

```python
# 篩選  NMOSD_v32 和 Control_v44 不相同的列
diff_genes_nmosd_control =merged_genes_nmosd_control[merged_genes_nmosd_control["HugoSymbol_v44"] != merged_genes_nmosd_control["HugoSymbol_v32"]]

print(diff_genes_nmosd_control.shape)
print(diff_genes_nmosd_control)
```

```python
# Check merge_df "Nan"
print("Merge nmosd_control ref genes 'Nan counts':\n",merged_genes_nmosd_control.isna().sum())
```

```python
diff_genes_nmosd_control.to_csv("./Dataset/ref_nmosd_control_different_genes.csv", index=False)
```

```python
merged_genes_nmosd_control
```

```python
#ENSG 重複 ->取第一筆
merge_dup_ENSG=merged_genes_nmosd_control[merged_genes_nmosd_control["ENSG"].duplicated(keep=False)]
print(f'Merge duplicated ENSG counts:',merge_dup_ENSG.shape[0])
merged_unique_ENSG = merged_genes_nmosd_control.drop_duplicates(subset="ENSG", keep="first")
merged_unique_ENSG= merged_unique_ENSG.loc[:,["ENSG","HugoSymbol_v44"]]
```
