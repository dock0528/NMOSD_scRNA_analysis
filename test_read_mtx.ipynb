```python
import scanpy as sc
import pandas as pd
import scipy.io as sio
```

```python
# matrix.mtx
X = sio.mmread("../scRNA_DATA/mtx_nmosd/NMOSD_matrix.mtx").tocsr().T  
# R 輸出是 genes × cells
# Scanpy/AnnData 需要 cells × genes
#.tocsr() 轉成稀疏矩陣

# Gene name (設為column)
var_names = pd.read_csv("../scRNA_DATA/mtx_nmosd/NMOSD_features.tsv", header=None)[0].values

# Cell name (設為row)
obs_names = pd.read_csv("../scRNA_DATA//mtx_nmosd/NMOSD_barcodes.tsv", header=None)[0].values

# 建立 AnnData
adata = sc.AnnData(X=X)
adata.var_names = var_names
adata.obs_names = obs_names

# metadata
obs = pd.read_csv("../scRNA_DATA//mtx_nmosd/NMOSD_obs.csv", index_col=0)
adata.obs = obs

```
