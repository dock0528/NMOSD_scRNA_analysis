```python
import scanpy as sc
import pandas as pd
import os
```

```python
########################【修改原本feature.tsv】########################
#先複製 mtx_merge 的資料夾
# === 1️⃣ 讀取原始單欄 features ===
features = pd.read_csv(f"/staging/biology/jane0528/NMOSD/scRNA/Dataset/My_merged_protein_coding_genes/My_mtx_merge_protein_coding/features.tsv", header=None, sep="\t")
genes = features[0].tolist()

# === 2️⃣ 建立三欄假格式 ===
fake_features = pd.DataFrame({
    0: [f"ENSG{i:09d}" for i in range(1, len(genes)+1)],  # 假的 gene_id
    1: genes,                                                   # 原本基因 symbol
    2: ["Gene Expression"] * len(genes)                         # 固定值
})

# === 3️⃣ 輸出新的 features 檔案 ===
fake_path = os.path.join("/staging/biology/jane0528/NMOSD/scRNA/Dataset/My_merged_protein_coding_genes/My_mtx_merge_protein_coding", "features_fixed.tsv")
fake_features.to_csv(fake_path, header=False, index=False, sep="\t") #記得改名為feature.tsv.gz!
```

```python
import scanpy as sc
import pandas as pd

#讀取原始 matrix.mtx 
#!!!!這裡的feature.tsv.gz 的第1&3欄是假的，只有第2欄是真的 >>> 為了讓它可以讓10x讀取
adata_raw = sc.read_10x_mtx(
    "/staging/biology/jane0528/NMOSD/scRNA/Dataset/My_merged_protein_coding_genes/My_mtx_merge_protein_coding",   # 包含 matrix.mtx、barcodes.tsv、features.tsv
    var_names="gene_symbols", #feature取第2欄
    cache=True
)
```

```python
# 讀入 top2000 HVG 的 .h5ad 
adata_hvg = sc.read_h5ad("/staging/biology/jane0528/NMOSD/scRNA/Dataset/My_merged_protein_coding_genes/Adata/My_merge_PCA_2000HVG(Leiden_pc9).h5ad")
```

```python
#----選出 top2000 基因 
common_genes = adata_raw.var_names.intersection(adata_hvg.var_names)
print("Common genes between raw and HVG:", len(common_genes))
```

```python
#----選出 top2000 基因 
common_genes = adata_raw.var_names.intersection(adata_hvg.var_names)
adata_sub = adata_raw[:, common_genes].copy()

# ----把標準化數值放進主矩陣 X 
adata_sub.X = adata_hvg[:, common_genes].X.copy()

# ----把原始 counts 保留到 layers["counts"]
adata_sub.layers["counts"] = adata_raw[:, common_genes].X.copy()

#若有 meta 資料（例如 cluster, Condition）可同步帶入
adata_sub.obs = adata_hvg.obs.copy()
```
